4.1. Adopted SO Methods(Algorithms): 

Dijkstra’s Algorithm: To determine the shortest route between two nodes in a graph, Dijkstra's algorithm is introduced. It is a well-known approach in the field of graph theory. It is regarded as one of the most effective algorithms for this purpose and was created by Edsger W. Dijkstra in 1956.

Starting at the source node, the algorithm iteratively explores nearby nodes to determine the shortest path to each one. The algorithm chooses the unvisited node that is the closest to the source node during each iteration and includes it in the collection of visited nodes. Then, if a shorter path to any of the selected node's neighbours is discovered, it looks at each of them individually and updates their distances. Until the final node is reached or all nodes have been visited, this procedure keeps going.

4.1.1. Steps involved in Dijkstra’s Algorithm:

•	 On the last vertex, mark a zero-distance. Make the current vertex this one. 
•	Compile a list of all vertex connections that lead to the present vertex. Determine their distance from the goal. We only need to add the most recent edge because we already know how far the current vertex is from the finish. 
•	If this distance exceeds a previously recorded distance, do not record it. 
•	Place a visiting mark at the current vertex. We'll never see another vertex like this again.
•	Repetition of step 2 will result in the current vertex being the vertex with the least distance.

### Creating class for Node, Graph, Vertex and Dikstra's Algorithm

```python
class Node:
    ''' 
        Each vertex of the graph is represented by the Node class. 
        The value of the attribute represents the data that was saved.
        The vertices with which a connection exists are represented by the list of neighbours attribute. 
    '''

    def __init__(self, value, neighbors=None):
        self.value = value
        if neighbors is None:
            self.neighbors = []
        else:
            self.neighbors = neighbors

    ''' If the vertex is related to at least one additional vertex, return True; otherwise, return false. '''

    def hs_neighbrz(self):
        if len(self.neighbors) == 0:
            return False
        return True

    ''' Returns the amount of vertices with connections, in '''


    def num_f_neighbrz(self):
        return len(self.neighbors)

    ''' Creates a new link and adds it to the neighbourhood list '''

    def ad_neibr(self, neighboor):
        self.neighbors.append(neighboor)

    def gt_neibrz(self):
        adjacent_node = []
        for n in self.neighbors:
            adjacent_node.append(n[0].value)
        return adjacent_node
    
    def __eq__(self, other):
        return self.value == other

    def __str__(self):
        retrnd_str = f"{self.value} -> "
        if self.hs_neighbrz():
            for neighboor in self.neighbors:
                retrnd_str += f"{neighboor[0].value} -> "

        retrnd_str += "None"
        return retrnd_str

class Graph:
    '''
     The graph data structure is represented by the Graph class.
        It includes a nodes attribute (list) that contains all of the graph's nodes.
    '''
    def __init__(self, nodes=None):
        if nodes is None:
            self.nodes = []
        else:
            self.nodes = nodes


    ''' Ad a new node in the grpah'''
    def ad_nw_node(self, node):
        self.nodes.append(node)


    '''If the node with the specified value exists, return True. If not, False is returned '''
    def fnd_nxt_node(self, value):
        for node in self.nodes:
            if node.value == value:
                return node
        return None


    '''Add a new edge between 2 nodes'''
    def ad_nw_edge(self, value1, value2, weight=1):
        node1 = self.fnd_nxt_node(value1)
        node2 = self.fnd_nxt_node(value2)

        if (node1 is not None) and (node2 is not None):
            node1.ad_neibr((node2, weight))
            node2.ad_neibr((node1, weight))
        else:
            print("Error: One or more nodes were not found")


    '''Return the graph's number of nodes'''

    def number_of_nodes(self):
        return f"The graph has {len(self.nodes)} nodes"
	

    ''' If the specified nodes are connected, return True. If not, return false '''
    def nodes_r_linked(self, node_one, node_two):
        node_one = self.fnd_nxt_node(node_one)
        node_two = self.fnd_nxt_node(node_two)

        for neighboor in node_one.neighbors:
            if neighboor[0].value == node_two.value:
                return True
        return False


    ''' Print the nodes '''
    def __str__(self):
        graph = ""
        for node in self.nodes:
            graph += f"{node.__str__()}\n"
        return graph

import math

''''Each vertex in a graph is represented by the class Vertex, which extends the class Node '''



class Vertex(Node):
    def __init__(self, value, neighbors=None):
        super().__init__(value, neighbors)
        self.previous_node = None
        self.visited = False

    ''' Return the difference from a given neighbor'''

    def dist_frm_neibr(self, node):
        for neighbor in self.neighbors:
            if neighbor[0].value == node.value:
                return neighbor[1]
        return None

    def __str__(self):
        return f"{self.value} {self.visited}"


''' Dijkstra Algorithm '''


class Dijkstra:
    def __init__(self, graph, start, target):
        self.graph = graph
        self.start = start
        self.target = target
        self.initialztn()

    ''' Initialize the labels of every vertex '''

    def initialztn(self):
        for node in self.graph.nodes:
            if node == self.start:
                node.length_from_start = 0

    ''' Calculate the return node with the highest transmission rate from the source node '''

    def retruned_node_selectd(self,value):
        next_node = None
                        
        return self.graph.fnd_nxt_node(value)

    '''  The algorithm's main. Put Dijkstra's methodical steps into practise '''

    def to_b_executd(self):
        src_node = self.graph.fnd_nxt_node(self.target)
        sourc_nod = src_node.value
        trgt_nod = self.start
        way_list = [[sourc_nod]]
        way_indx = 0
        prv_nodes = [sourc_nod]
        crnt_way =[]
        nxt_nodz = src_node.gt_neibrz()
        
        
        #If source and destination are similar
        if sourc_nod == trgt_nod:
            return way_list[0]
        
        while way_indx < len(way_list):
        
            # Select the node with the minimun distrance from beginning positi
            
            crnt_way = way_list[way_indx]
            last_node = crnt_way[-1]
            select_node = self.retruned_node_selectd(last_node)
            nxt_nodz = select_node.gt_neibrz()  

            
            # Search aim node
            if trgt_nod in nxt_nodz:
                crnt_way.append(trgt_nod)
                return crnt_way, (len(crnt_way)-1) * 30
            
            # Add new paths
            for next_node in nxt_nodz:
                if not next_node in prv_nodes:
                    new_path = crnt_way[:]
                    new_path.append(next_node)
                    way_list.append(new_path) 
                    
                    # To avoid backtracking
                    prv_nodes.append(next_node)

            # Continue to next path in list
            way_indx += 1

            # Determine the route between the source and the target nodes.

       
        return [], 0
```

```python
# Read flights details like Flight number, Altitude, Latitude, Longitude from the CSV file
Fltlistdata = pd.read_csv("/content/sample_data/NA_11_Jun_29_2018_UTC11.CSV")

# Set ground station area with (Latitude, Longitude, Altitude)
GS_LHR = [51.47, -0.45, 24.91]    
GS_EWR = [40.69, -74.17, 2.66]
GS_UAK = [61.9, -41.25, 34.36]  

# The radius of earth
RadEr = 6371000 

# Conversion of given (latitude, longitude, altitude) value 
# to its corresponding cartisian coordinates
def cnvrsn_to_car_coordintes(node):
    longitude = np.deg2rad(Fltlistdata._get_value(node, 'Longitude'))
    latitude = np.deg2rad(Fltlistdata._get_value(node, 'Latitude'))
    altitude = Fltlistdata._get_value(node, 'Altitude') * 0.304

    X_co = (RadEr + altitude)* math.cos(latitude) * math.cos(longitude) 
    Y_co = (RadEr + altitude)* math.cos(latitude) * math.sin(longitude) 
    Z_co = (RadEr + altitude)* math.sin(latitude) 
    
    return(X_co,Y_co,Z_co)
```

```python
# Cartisian Coordinates for the ground stations - Heathrow, Newark
latitude = np.deg2rad(GS_LHR[0]) 
longitude =  np.deg2rad(GS_LHR[1]) 
lhrgs_x = (RadEr + GS_LHR[2])* math.cos(latitude) * math.cos(longitude) 
lhrgs_y = (RadEr + GS_LHR[2])* math.cos(latitude) * math.sin(longitude) 
lhrgs_z = (RadEr + GS_LHR[2])* math.sin(latitude) 
 
latitude = np.deg2rad(GS_EWR[0]) 
longitude =  np.deg2rad(GS_EWR[1]) 
ewrgs_x = (RadEr + GS_EWR[2])* math.cos(latitude) * math.cos(longitude) 
ewrgs_y = (RadEr + GS_EWR[2])* math.cos(latitude) * math.sin(longitude) 
ewrgs_z = (RadEr + GS_EWR[2])* math.sin(latitude)

latitude = np.deg2rad(GS_UAK[0]) 
longitude =  np.deg2rad(GS_UAK[1]) 
uakgs_x = (RadEr + GS_UAK[2])* math.cos(latitude) * math.cos(longitude) 
uakgs_y = (RadEr + GS_UAK[2])* math.cos(latitude) * math.sin(longitude) 
uakgs_z = (RadEr + GS_UAK[2])* math.sin(latitude)

# Coordinates of Heathrow and Newark ground stations
Htrw_coor = [lhrgs_x, lhrgs_y, lhrgs_z]
Nwrk_coor = [ewrgs_x, ewrgs_y, ewrgs_z]
Nsrq_coor = [uakgs_x, uakgs_y, uakgs_z]

# Initializing value for x,y,z coordinates for dataframe
Fltlistdata['X_co'] = 0
Fltlistdata['Y_co'] = 0
Fltlistdata['Z_co'] = 0

# Converting coordinates of flights in file to cartesian coordinates
for flight, row in Fltlistdata.iterrows():
    xcoord,ycoord,zcoord = cnvrsn_to_car_coordintes(flight)
    Fltlistdata['X_co'][flight] = xcoord
    Fltlistdata['Y_co'][flight] = ycoord
    Fltlistdata['Z_co'][flight] = zcoord

# Appending cartisian coordinates of Heathrow and Newark to Flight Details
Fltlistdata = Fltlistdata.append({ 'Flight': 'LHR', 
                    'X_co': lhrgs_x, 'Y_co': lhrgs_y, 
                    'Z_co': lhrgs_z}, ignore_index = True)
Fltlistdata = Fltlistdata.append({ 'Flight': 'EWR', 
                    'X_co': ewrgs_x, 'Y_co': ewrgs_y, 
                    'Z_co': ewrgs_z}, ignore_index = True)

ID_lhrgs= len(Fltlistdata)-3
ID_ewrgs = len(Fltlistdata)-2
ID_uakgs = len(Fltlistdata)-1
```

```python
# Calculate the difference between two nodes, 
# either flight or ground station
def Spanbwtwopints(node1, node2):   
    return math.sqrt((node1[0]-node2[0])**2 + (node1[1]-node2[1])**2 + 
                     (node1[2]-node2[2])**2)/1000

# Find cartesian coordinate of flight or ground station 
# at a particular index
def CartesianCoordinates(index):
    return [Fltlistdata.loc[index, 'X_co'], 
            Fltlistdata.loc[index, 'Y_co'], 
            Fltlistdata.loc[index, 'Z_co']]

# Distance and related switching thresholds are given 
# transmission rates
def findlimofdaratransmission(nodeGap):
    dataRate = 0
    if nodeGap > 550: 
        dataRate = 11.023
    elif nodeGap > 350:  
        dataRate = 31.895
    elif nodeGap > 210:  
        dataRate = 43.031
    elif nodeGap > 130:  
        dataRate = 52.857
    elif nodeGap > 90:  
        dataRate = 66.048
    elif nodeGap > 35:  
        dataRate = 77.071
    else: 
        dataRate = 99.275
    return dataRate

# Calculate the difference between 2 flights or/and ground station
# given the vertices of the flights
def spanbwtwopointz(flight1, flight2):
    return Spanbwtwopints(
                CartesianCoordinates(flight1), 
                CartesianCoordinates(flight2))
```

```python
# find best neighbor of current node from all posibilities
def Find_bstneighbr(neighbors, visited):
    bst_neighbr_indx = np.argmax(neighbors)
    bst_neighbr_val = np.amax(neighbors)
    if(bst_neighbr_indx in visited):
        neighbors[bst_neighbr_indx] = 0
        bst_neighbr_indx, bst_neighbr_val = Find_bstneighbr(neighbors, visited)
    return bst_neighbr_indx, bst_neighbr_val

# Find optimized path between flights and ground station using Dijkstra Algorithm
def dijkstra(graph):
    visited = set()
    unvisited = list(range(len(graph)-2))
    routes = []
    rates = []
    while len(unvisited)>0:
        localRate = 400
        frst_indx = random.choice(unvisited)
        current = frst_indx
        traversedNodes = [current]
        visited.add(current)
        unvisited.remove(current)
        count = 1
        while current not in [ID_lhrgs,ID_ewrgs] and (count < 4):
            current, current_data_rate = Find_bstneighbr(graph[current], visited)
            localRate = current_data_rate if localRate > current_data_rate else localRate
            traversedNodes.append(current)
            count += 1
            if current not in [ID_lhrgs,ID_ewrgs]:
                visited.add(current)
                unvisited.remove(current)
            else:
                break
        if current not in [ID_lhrgs,ID_ewrgs] and count == 4:
            d1=spanbwtwopointz(traversedNodes[-1],ID_lhrgs)
            d2=spanbwtwopointz(traversedNodes[-1],ID_ewrgs)
            if(d1>d2):
                traversedNodes.append(ID_ewrgs)
            else:
                traversedNodes.append(ID_lhrgs)
        routes.append(traversedNodes)
        rates.append(localRate)
    return routes, rates
```

#6. Validation & Evaluation through comparison

6.1. Benefits of Dijkstra’s Algorithm in Aeronautical Optimisation:

Quickness: Particularly for dense graphs, the Dijkstra algorithm is fairly quick. In aeronautical optimisation issues, which usually include huge networks with numerous nodes and edges, Dijkstra's method can quickly identify the best solution.

Straight forward: Dijkstra's method is straightforward to apply, which makes it simple to comprehend and adapt to particular aviation optimisation challenges. 

Versatility: Dijkstra's method is effective at solving both the single-pair shortest path problem and the all-pairs shortest path problem, both of which are significant in aeronautical optimisation. 

Optimality: Dijkstra's technique makes sure that the shortest path between any two nodes is found in scenarios requiring aeronautical optimisation, where the goal is frequently to reduce the distance or trip time.

6.2. Drawbacks of Dijkstra’s in optimisation:

•	Because of high temporal complexity, it performs less well for big graphs. 
•	It calls for the use of a priority queue, which might be tricky to implement in some languages. 
•	The method does not work with negative weighted graphs. 
•	 It struggles to handle graphs with many sources or destinations. 
•	 It performs poorly with graphs that contain a lot of vertices since it must process each one separately. It is not a good fit for distributed or parallel computation.
